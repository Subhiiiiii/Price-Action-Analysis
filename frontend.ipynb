```python
# =====================
# Paste each cell into your Jupyter notebook in order.
# Fetches OHLC, scores, labels, then visualises all regime features.

# Works with your existing market_analysis package as-is.
# No changes to calculate.py / label.py needed.
# """

# ═══════════════════════════════════════════════════════════════
# CELL 1 — imports & config
# ═══════════════════════════════════════════════════════════════

import sys
sys.path.insert(0, r"C:\\Users\\shresth.shankhdhar\\OneDrive - hertshtengroup.com\\Documents")

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from scipy.stats import gaussian_kde

from market_analysis.data.ohlc import getData_OHLC_v2, get_all_horizon_ohlc
from market_analysis.signals.absorption.calculate import calculate_absorption
from market_analysis.signals.absorption.label     import label_absorption
from market_analysis.utils.contracts              import get_all_front_instruments
from market_analysis.config                       import (
    PRODUCTS, REFERENCE_DATE, ABSORPTION, FRONT_N
)

plt.style.use("dark_background")
matplotlib.rcParams["figure.dpi"] = 120
print("Imports OK")


# ═══════════════════════════════════════════════════════════════
# CELL 2 — fetch OHLC + score + label for all instruments
# ═══════════════════════════════════════════════════════════════

PRODUCT_CODE = "SM"
HORIZONS     = ["1d", "2d", "3d", "5d"]
TICK         = PRODUCTS[PRODUCT_CODE]["tick_size"]     # 0.1
CFG          = ABSORPTION

# ── get instrument list ──────────────────────────────────────
by_struct = get_all_front_instruments(PRODUCT_CODE, REFERENCE_DATE, FRONT_N)
print("Instruments:")
for s, instrs in by_struct.items():
    print(f"  {s}: {instrs}")

# ── fetch + score + label every instrument × horizon ────────
all_labeled: dict[str, dict[str, pd.DataFrame]] = {}

for struct_name, instruments in by_struct.items():
    for instr in instruments:
        print(f"\\n  [{struct_name}] {instr} ...", end=" ")

        # fetch 1D once, build all horizons from it
        horizon_ohlc = get_all_horizon_ohlc(
            instrument     = instr,
            product_code   = PRODUCT_CODE,
            horizons       = HORIZONS,
            getData_OHLC_v2= getData_OHLC_v2,
        )
        if not horizon_ohlc:
            print("NO DATA — skip")
            continue

        all_labeled[instr] = {}

        for horizon, ohlc in horizon_ohlc.items():
            if ohlc is None or len(ohlc) < CFG["window"] + CFG["forward_days"] + 5:
                print(f"{horizon}:too_short ", end="")
                continue

            # ── score ────────────────────────────────────────
            scored = calculate_absorption(
                ohlc,
                window             = CFG["window"],
                tick_size          = TICK,
                strength_threshold = CFG["strength_threshold"],
            )

            # ── label (forward labels — for colouring only) ─
            labeled = label_absorption(
                scored,
                tick                = TICK,
                required_move_ticks = CFG["required_move_ticks"],
                sl_lookback         = CFG["sl_lookback"],
                forward_days        = CFG["forward_days"],
            )

            # keep only rows with a direction
            labeled = labeled[
                labeled["absorption_direction"].isin(["BUY", "SELL"])
            ].dropna(subset=["absorption_strength"]).copy()
            labeled["success"] = labeled["sl_success"]
            labeled["horizon"] = horizon
            labeled["instrument"] = instr
            labeled["struct"]   = struct_name

            all_labeled[instr][horizon] = labeled
            print(f"{horizon}:{len(labeled)} ", end="")

print("\\n\\nDone fetching.")

# ── quick count ─────────────────────────────────────────────
total = sum(len(df) for hd in all_labeled.values() for df in hd.values())
print(f"Total labeled rows across all instruments × horizons: {total}")

# ═══════════════════════════════════════════════════════════════
# CELL 3 — build enriched DataFrame (add microstructure features)
# ═══════════════════════════════════════════════════════════════

FW        = 5    # forward window for post-event features
VOL_WIN   = 20   # rolling window for realised vol

def enrich(df: pd.DataFrame, fw: int = 5, vol_win: int = 20) -> pd.DataFrame:
    # \"\"\"
    # Add post-event features to a labeled DataFrame.
    # All post-event columns are label-only (used for colouring/analysis,
    # never fed back into absorption_strength).
    # \"\"\"
    d = df.copy().sort_values("Date").reset_index(drop=True)
    n = len(d)

    close  = d["close"].to_numpy(float)
    delta  = d["delta"].to_numpy(float)
    high   = d["high"].to_numpy(float) if "high"  in d.columns else close.copy()
    low    = d["low"].to_numpy(float)  if "low"   in d.columns else close.copy()

    # 1. Kyle lambda (past-only)
    price_chg = np.abs(np.diff(close, prepend=np.nan))
    abs_delta = np.abs(delta)
    lam = price_chg / (abs_delta + 1e-9)
    d["kyle_lambda"] = lam

    # 2. Realised vol (rolling std of log-returns, past-only)
    ret  = np.diff(np.log(np.where(close > 0, close, np.nan)), prepend=np.nan)
    rvol = pd.Series(ret).rolling(vol_win, min_periods=3).std().to_numpy()
    d["rvol"] = rvol

    # 3. Absorption Efficiency  (normalised)
    roll_adelta = pd.Series(abs_delta).rolling(vol_win, min_periods=3).median().to_numpy()
    roll_rvol   = pd.Series(rvol).rolling(vol_win, min_periods=3).median().to_numpy()
    ae = (abs_delta / (roll_adelta + 1e-9)) / (price_chg / (roll_rvol + 1e-9) + 1e-9)
    d["absorption_efficiency"] = ae

    # 4. Post-event features  ← forward-looking, for colouring only
    delta_persist = np.full(n, np.nan)
    lambda_shift  = np.full(n, np.nan)
    delta_accel   = np.full(n, np.nan)
    fwd_price_chg = np.full(n, np.nan)

    delta_std = np.nanstd(delta) + 1e-9

    for i in range(n - fw):
        fw_delta = delta[i+1 : i+fw+1]
        fw_close = close[i+1 : i+fw+1]

        # delta persistence: fraction of forward bars with same-sign delta
        same = np.sum(np.sign(fw_delta) == np.sign(delta[i]))
        delta_persist[i] = same / fw

        # kyle lambda shift
        fw_pc  = np.abs(np.diff(np.concatenate([[close[i]], fw_close])))
        fw_lam = fw_pc / (np.abs(fw_delta) + 1e-9)
        if np.isfinite(lam[i]) and lam[i] > 1e-12:
            lambda_shift[i] = np.nanmean(fw_lam) / (lam[i] + 1e-9)

        # delta acceleration
        delta_accel[i] = (np.nanmean(fw_delta) - delta[i]) / delta_std

        # raw forward price change
        fwd_price_chg[i] = fw_close[-1] - close[i]

    d["delta_persist"]   = delta_persist
    d["lambda_shift"]    = lambda_shift
    d["delta_accel"]     = delta_accel
    d["fwd_price_chg"]   = fwd_price_chg

    # 5. Regime label (for colouring only — never a model input)
    ABS_THR = ABSORPTION["strength_threshold"]
    regime  = np.full(n, "UNRESOLVED", dtype=object)
    abs_str = d["absorption_strength"].to_numpy(float)
    dirn    = d["absorption_direction"].str.upper().to_numpy()
    has_lbl = "sl_success" in d.columns

    for i in range(n - fw):
        is_abs = np.isfinite(abs_str[i]) and abs_str[i] >= ABS_THR
        dir_i  = dirn[i]

        if not is_abs:
            if dir_i == "BUY"  and fwd_price_chg[i] > 0: regime[i] = "PURE_BUYING"
            elif dir_i == "SELL" and fwd_price_chg[i] < 0: regime[i] = "PURE_SELLING"
            continue

        if has_lbl and "sl_success" in d.columns:
            sl_val = d["sl_success"].iloc[i]
            if np.isfinite(sl_val):
                regime[i] = "ABS_CONT" if sl_val == 1 else "ABS_REV"
                continue

        # fallback to raw price direction
        if dir_i == "BUY":
            regime[i] = "ABS_REV"  if fwd_price_chg[i] < 0 else "ABS_CONT"
        else:
            regime[i] = "ABS_CONT" if fwd_price_chg[i] > 0 else "ABS_REV"

    d["regime"] = regime
    return d

# ── enrich all DataFrames ────────────────────────────────────
all_enriched: dict[str, dict[str, pd.DataFrame]] = {}
for instr, hd in all_labeled.items():
    all_enriched[instr] = {}
    for horizon, df in hd.items():
        if len(df) > FW + 5:
            all_enriched[instr][horizon] = enrich(df, fw=FW, vol_win=VOL_WIN)

print("Enrichment done.")

# ── merge one big DataFrame for cross-instrument analysis ───
big = pd.concat(
    [df for hd in all_enriched.values() for df in hd.values()],
    ignore_index=True,
)
print(f"Combined DataFrame: {big.shape}")
print(big[["instrument","horizon","regime","absorption_strength",
           "delta_persist","lambda_shift","delta_accel"]].head(10))


# ═══════════════════════════════════════════════════════════════
# CELL 4 — PLOT: regime count per instrument × horizon
# ═══════════════════════════════════════════════════════════════

# ── Regime count heatmap: absorption cases only ──────────────
abs_cases = big[big["regime"].isin(["ABS_REV","ABS_CONT"])].copy()

pivot = (
    abs_cases
    .groupby(["instrument","horizon","regime"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)
pivot["total"]    = pivot.get("ABS_REV", 0) + pivot.get("ABS_CONT", 0)
pivot["cont_pct"] = pivot.get("ABS_CONT", 0) / (pivot["total"] + 1e-9) * 100

print("\\n── Absorption Cases per Instrument × Horizon ─────────────")
print(pivot.to_string(index=False))

# ═══════════════════════════════════════════════════════════════
# CELL 5 — PLOT: full 6-panel regime explorer
# ═══════════════════════════════════════════════════════════════
HORIZON_TO_PLOT = "1d"   # change to "2d", "3d", "5d" as needed

# pool all instruments for this horizon
frames = [
    df for hd in all_enriched.values()
    for h, df in hd.items() if h == HORIZON_TO_PLOT
]
plot_df = pd.concat(frames, ignore_index=True) if frames else big.copy()

C_REV   = "#F44336"
C_CONT  = "#4CAF50"
C_BUY   = "#2196F3"
C_SELL  = "#FF9800"
C_NEUT  = "#9E9E9E"
ALPHA   = 0.55

REGIME_COLOR = {
    "PURE_BUYING":  C_BUY,
    "PURE_SELLING": C_SELL,
    "ABS_REV":      C_REV,
    "ABS_CONT":     C_CONT,
    "UNRESOLVED":   C_NEUT,
}

def rcolors(series):
    return [REGIME_COLOR.get(r, C_NEUT) for r in series]

fig = plt.figure(figsize=(20, 14), constrained_layout=True)
fig.suptitle(
    f"Absorption Regime Explorer  ▸  SM  ▸  horizon={HORIZON_TO_PLOT}  "
    f"▸  n={len(plot_df)} bars",
    fontsize=13, fontweight="bold"
)
gs = gridspec.GridSpec(3, 3, figure=fig)

# ── Panel 1: Pressure vs Response map (biggest panel) ───────
ax1 = fig.add_subplot(gs[0:2, 0:2])
d_ = plot_df.dropna(subset=["delta_pct","absorption_efficiency","regime"])
if len(d_) > 3000: d_ = d_.sample(3000, random_state=42)
x_ = d_["delta_pct"].clip(d_["delta_pct"].quantile(0.01),
                            d_["delta_pct"].quantile(0.99))
y_ = np.log1p(d_["absorption_efficiency"].clip(0))
ax1.scatter(x_, y_, c=rcolors(d_["regime"]), alpha=ALPHA, s=18, linewidths=0)
ax1.axvline(0, color="white", lw=0.8, alpha=0.4)
ax1.axhline(np.log1p(1), color="white", lw=0.8, ls="--", alpha=0.4)
ax1.set_xlabel("Delta %  (signed,  + = net buying)", fontsize=10)
ax1.set_ylabel("log(1 + Absorption Efficiency)", fontsize=10)
ax1.set_title("Pressure vs Response Map", fontsize=11, fontweight="bold")
from matplotlib.patches import Patch
handles = [Patch(fc=v, label=k, alpha=0.85) for k, v in REGIME_COLOR.items()]
ax1.legend(handles=handles, fontsize=8, loc="upper left")
ax1.set_facecolor("#111")

# ── Panel 2: Delta Persistence KDE ──────────────────────────
ax2 = fig.add_subplot(gs[0, 2])
for regime, label, color in [("ABS_REV","Abs→Reversal",C_REV),
                               ("ABS_CONT","Abs→Continuation",C_CONT)]:
    arr = plot_df.loc[plot_df["regime"]==regime,"delta_persist"].dropna().to_numpy()
    if len(arr) < 4: continue
    xs = np.linspace(0, 1, 200)
    try:
        ys = gaussian_kde(arr, bw_method="scott")(xs)
        ax2.fill_between(xs, ys, alpha=0.35, color=color)
        ax2.plot(xs, ys, color=color, lw=2, label=f"{label} (n={len(arr)})")
    except Exception:
        ax2.hist(arr, bins=15, color=color, alpha=0.5, density=True, label=label)
ax2.set_xlabel("Delta Persistence", fontsize=9)
ax2.set_ylabel("Density", fontsize=9)
ax2.set_title("Delta Persistence:\\nReversal vs Continuation", fontsize=10, fontweight="bold")
ax2.legend(fontsize=8)
ax2.set_facecolor("#111")

# ── Panel 3: Kyle Lambda Shift KDE ──────────────────────────
ax3 = fig.add_subplot(gs[1, 2])
d_ls = plot_df.dropna(subset=["lambda_shift"])
d_ls = d_ls[np.isfinite(d_ls["lambda_shift"]) & (d_ls["lambda_shift"] > 0)]
for regime, label, color in [("ABS_REV","Abs→Reversal",C_REV),
                               ("ABS_CONT","Abs→Continuation",C_CONT)]:
    arr = np.log(d_ls.loc[d_ls["regime"]==regime,"lambda_shift"].to_numpy() + 1e-9)
    arr = arr[np.isfinite(arr)]
    if len(arr) < 4: continue
    xs = np.linspace(arr.min(), arr.max(), 200)
    try:
        ys = gaussian_kde(arr, bw_method="scott")(xs)
        ax3.fill_between(xs, ys, alpha=0.35, color=color)
        ax3.plot(xs, ys, color=color, lw=2, label=f"{label} (n={len(arr)})")
    except Exception:
        ax3.hist(arr, bins=15, color=color, alpha=0.5, density=True, label=label)
ax3.axvline(0, color="white", lw=1, ls="--", alpha=0.6, label="λ shift = 1")
ax3.set_xlabel("log(λ_post / λ_absorption)", fontsize=9)
ax3.set_ylabel("Density", fontsize=9)
ax3.set_title("Kyle λ Shift after Absorption", fontsize=10, fontweight="bold")
ax3.legend(fontsize=8)
ax3.set_facecolor("#111")

# ── Panel 4: Delta Acceleration scatter ─────────────────────
ax4 = fig.add_subplot(gs[2, 0])
d_abs = plot_df[
    plot_df["absorption_strength"] >= ABSORPTION["strength_threshold"]
].dropna(subset=["delta_accel","absorption_strength"])
ax4.scatter(
    d_abs["absorption_strength"],
    d_abs["delta_accel"].clip(-5, 5),
    c=rcolors(d_abs["regime"]), alpha=ALPHA, s=18, linewidths=0,
)
ax4.axhline(0, color="white", lw=1, ls="--", alpha=0.5)
ax4.set_xlabel("Absorption Strength", fontsize=9)
ax4.set_ylabel("Delta Acceleration (normalised)", fontsize=9)
ax4.set_title("Delta Acceleration\\nvs Absorption Strength", fontsize=10, fontweight="bold")
ax4.legend(handles=[Patch(fc=C_REV,label="Abs→Rev",alpha=0.8),
                     Patch(fc=C_CONT,label="Abs→Cont",alpha=0.8)], fontsize=8)
ax4.set_facecolor("#111")

# ── Panel 5: Absorption Efficiency vs Realised Vol ──────────
ax5 = fig.add_subplot(gs[2, 1])
d_ae = plot_df.dropna(subset=["absorption_efficiency","rvol","regime"])
d_ae = d_ae[np.isfinite(d_ae["absorption_efficiency"]) & (d_ae["rvol"] > 0)]
if len(d_ae) > 2000: d_ae = d_ae.sample(2000, random_state=42)
ax5.scatter(
    d_ae["rvol"],
    np.log1p(d_ae["absorption_efficiency"].clip(0)),
    c=rcolors(d_ae["regime"]), alpha=ALPHA, s=16, linewidths=0,
)
ax5.set_xlabel("Realised Volatility", fontsize=9)
ax5.set_ylabel("log(1 + AE)", fontsize=9)
ax5.set_title("Absorption Efficiency\\nvs Market Volatility", fontsize=10, fontweight="bold")
ax5.legend(handles=handles, fontsize=7)
ax5.set_facecolor("#111")

# ── Panel 6: Reversal vs Continuation by absorption class ───
ax6 = fig.add_subplot(gs[2, 2])
d_cls = plot_df[plot_df["regime"].isin(["ABS_REV","ABS_CONT"])]
if "absorption_class" in d_cls.columns and len(d_cls) > 0:
    tbl   = d_cls.groupby(["absorption_class","regime"]).size().unstack(fill_value=0)
    order = [o for o in ["POSSIBLE","MODERATE","STRONG","EXTREME"] if o in tbl.index]
    tbl   = tbl.reindex(order)
    total = tbl.sum(axis=1)
    frac  = tbl.div(total, axis=0)
    x     = np.arange(len(frac))
    bottom= np.zeros(len(frac))
    for col, color, lbl in [("ABS_REV",C_REV,"Abs→Rev"),("ABS_CONT",C_CONT,"Abs→Cont")]:
        if col in frac.columns:
            vals = frac[col].to_numpy()
            ax6.bar(x, vals, 0.55, bottom=bottom, color=color, alpha=0.85, label=lbl)
            bottom += vals
    ax6.set_xticks(x)
    ax6.set_xticklabels(frac.index, fontsize=9)
    for xi, (idx, _) in enumerate(frac.iterrows()):
        ax6.text(xi, 1.03, f"n={int(total.loc[idx])}", ha="center",
                 fontsize=8, color="white")
    ax6.set_ylim(0, 1.15)
    ax6.set_xlabel("Absorption Class", fontsize=9)
    ax6.set_ylabel("Fraction", fontsize=9)
    ax6.set_title("Reversal vs Continuation\\nby Absorption Class", fontsize=10, fontweight="bold")
    ax6.legend(fontsize=8)
ax6.set_facecolor("#111")

plt.show()


# ═══════════════════════════════════════════════════════════════
# CELL 6 — per-instrument summary table
# ═══════════════════════════════════════════════════════════════

rows = []
for instr, hd in all_enriched.items():
    for horizon, df in hd.items():
        for regime in ["PURE_BUYING","PURE_SELLING","ABS_REV","ABS_CONT"]:
            sub = df[df["regime"] == regime]
            if len(sub) == 0: continue
            rows.append({
                "Instrument" : instr,
                "Horizon"    : horizon,
                "Regime"     : regime,
                "Count"      : len(sub),
                "Mean AStr"  : round(sub["absorption_strength"].mean(), 3),
                "Mean AE"    : round(sub["absorption_efficiency"].mean(), 2),
                "Mean DP"    : round(sub["delta_persist"].mean(), 2),
                "Mean λShift": round(sub["lambda_shift"].mean(), 2),
                "Mean Δaccel": round(sub["delta_accel"].mean(), 2),
                "Mean FwdΔP" : round(sub["fwd_price_chg"].mean(), 4),
            })

summary = pd.DataFrame(rows)
print(summary.to_string(index=False))


if __name__ == "__main__":
    print("This file contains notebook cell templates.")
    print("Copy each CELL_N string into a Jupyter cell and run in order.")
    for i in range(1, 7):
        print(f"\n{'='*60}")
        print(f"CELL {i}")
        print('='*60)
        g = globals()[f"CELL_{i}"]
        print(g)
```

```python
# ═══════════════════════════════════════════════════════════════
# CELL 3 — ENRICH WITH PAST-ONLY MICROSTRUCTURE FEATURES
# ═══════════════════════════════════════════════════════════════

FW      = 5
VOL_WIN = 20


# ─────────────────────────────────────────────────────────────
# 1. PAST-ONLY DELTA PERSISTENCE
# ─────────────────────────────────────────────────────────────
#
# DP_N(t) =
#   fraction of previous N bars whose delta
#   has the same sign as delta[t]
#
# IMPORTANT:
# Uses ONLY t-N ... t-1
# Never uses t+1, t+2, ...
# ─────────────────────────────────────────────────────────────

def add_past_delta_persistence(d, lookbacks=(3, 5, 10)):

    delta = d["delta"].to_numpy(float)
    n = len(d)

    for lookback in lookbacks:

        result = np.full(n, np.nan)

        for i in range(lookback, n):

            current_delta = delta[i]

            if not np.isfinite(current_delta) or current_delta == 0:
                continue

            past = delta[i-lookback:i]

            valid = np.isfinite(past) & (past != 0)

            if valid.sum() == 0:
                continue

            result[i] = np.mean(
                np.sign(past[valid]) == np.sign(current_delta)
            )

        d[f"delta_persist_past_{lookback}"] = result

    return d


# ─────────────────────────────────────────────────────────────
# 2. CUMULATIVE DELTA PRESSURE — PAST ONLY
# ─────────────────────────────────────────────────────────────
#
# CDP_N =
#       sum(previous N deltas)
#       ---------------------
#       sum(abs(previous N deltas))
#
# Range:
#       -1 <= CDP <= +1
#
# +1  = almost entirely buying pressure
# -1  = almost entirely selling pressure
#  0  = balanced
#
# IMPORTANT:
# Uses ONLY previous N bars.
# ─────────────────────────────────────────────────────────────

def add_cumulative_delta_pressure(d, lookbacks=(3, 5, 10)):

    delta = d["delta"].to_numpy(float)
    n = len(d)

    for lookback in lookbacks:

        result = np.full(n, np.nan)

        for i in range(lookback, n):

            past = delta[i-lookback:i]

            valid = np.isfinite(past)

            if valid.sum() == 0:
                continue

            past = past[valid]

            denominator = np.sum(np.abs(past))

            if denominator <= 1e-12:
                continue

            numerator = np.sum(past)

            result[i] = numerator / denominator

        d[f"cumulative_delta_pressure_{lookback}"] = result

    return d


# ─────────────────────────────────────────────────────────────
# 3. MAIN ENRICH FUNCTION
# ─────────────────────────────────────────────────────────────

def enrich(df: pd.DataFrame,
           fw: int = FW,
           vol_win: int = VOL_WIN) -> pd.DataFrame:

    d = df.copy().sort_values("Date").reset_index(drop=True)

    n = len(d)

    close = d["close"].to_numpy(float)
    delta = d["delta"].to_numpy(float)

    high = (
        d["high"].to_numpy(float)
        if "high" in d.columns
        else close.copy()
    )

    low = (
        d["low"].to_numpy(float)
        if "low" in d.columns
        else close.copy()
    )


    # ═══════════════════════════════════════════════════════════
    # A. CURRENT-BAR KYLE LAMBDA
    # ═══════════════════════════════════════════════════════════

    price_chg = np.abs(
        np.diff(close, prepend=np.nan)
    )

    abs_delta = np.abs(delta)

    lam = price_chg / (abs_delta + 1e-9)

    d["kyle_lambda"] = lam


    # ═══════════════════════════════════════════════════════════
    # B. REALISED VOLATILITY
    # ═══════════════════════════════════════════════════════════

    ret = np.diff(
        np.log(
            np.where(close > 0, close, np.nan)
        ),
        prepend=np.nan
    )

    rvol = (
        pd.Series(ret)
        .rolling(vol_win, min_periods=3)
        .std()
        .to_numpy()
    )

    d["rvol"] = rvol


    # ═══════════════════════════════════════════════════════════
    # C. ABSORPTION EFFICIENCY
    # ═══════════════════════════════════════════════════════════

    roll_adelta = (
        pd.Series(abs_delta)
        .rolling(vol_win, min_periods=3)
        .median()
        .to_numpy()
    )

    roll_rvol = (
        pd.Series(rvol)
        .rolling(vol_win, min_periods=3)
        .median()
        .to_numpy()
    )

    ae = (
        (abs_delta / (roll_adelta + 1e-9))
        /
        (
            price_chg / (roll_rvol + 1e-9)
            + 1e-9
        )
    )

    d["absorption_efficiency"] = ae


    # ═══════════════════════════════════════════════════════════
    # D. PAST-ONLY FEATURES
    # ═══════════════════════════════════════════════════════════

    d = add_past_delta_persistence(
        d,
        lookbacks=(3, 5, 10)
    )

    d = add_cumulative_delta_pressure(
        d,
        lookbacks=(3, 5, 10)
    )


    # ═══════════════════════════════════════════════════════════
    # E. FUTURE FEATURES
    #
    # These are ONLY for analysing what happened AFTER
    # absorption. DO NOT use these as live model inputs.
    # ═══════════════════════════════════════════════════════════

    delta_persist = np.full(n, np.nan)
    lambda_shift  = np.full(n, np.nan)
    delta_accel   = np.full(n, np.nan)
    fwd_price_chg = np.full(n, np.nan)

    delta_std = np.nanstd(delta) + 1e-9

    for i in range(n - fw):

        fw_delta = delta[i+1:i+fw+1]

        fw_close = close[i+1:i+fw+1]


        # Future delta persistence
        same = np.sum(
            np.sign(fw_delta) == np.sign(delta[i])
        )

        delta_persist[i] = same / fw


        # Future Kyle lambda
        fw_pc = np.abs(
            np.diff(
                np.concatenate([[close[i]], fw_close])
            )
        )

        fw_lam = (
            fw_pc /
            (np.abs(fw_delta) + 1e-9)
        )

        if np.isfinite(lam[i]) and lam[i] > 1e-12:

            lambda_shift[i] = (
                np.nanmean(fw_lam)
                /
                (lam[i] + 1e-9)
            )


        # Future delta acceleration
        delta_accel[i] = (
            np.nanmean(fw_delta) - delta[i]
        ) / delta_std


        # Future price movement
        fwd_price_chg[i] = (
            fw_close[-1] - close[i]
        )


    d["delta_persist"] = delta_persist
    d["lambda_shift"]  = lambda_shift
    d["delta_accel"]   = delta_accel
    d["fwd_price_chg"] = fwd_price_chg


    # ═══════════════════════════════════════════════════════════
    # F. REGIME LABEL
    # ═══════════════════════════════════════════════════════════

    ABS_THR = ABSORPTION["strength_threshold"]

    regime = np.full(
        n,
        "UNRESOLVED",
        dtype=object
    )

    abs_str = d[
        "absorption_strength"
    ].to_numpy(float)

    dirn = (
        d["absorption_direction"]
        .str.upper()
        .to_numpy()
    )

    has_lbl = "sl_success" in d.columns


    for i in range(n - fw):

        is_abs = (
            np.isfinite(abs_str[i])
            and abs_str[i] >= ABS_THR
        )

        dir_i = dirn[i]


        if not is_abs:

            if (
                dir_i == "BUY"
                and fwd_price_chg[i] > 0
            ):
                regime[i] = "PURE_BUYING"

            elif (
                dir_i == "SELL"
                and fwd_price_chg[i] < 0
            ):
                regime[i] = "PURE_SELLING"

            continue


        if has_lbl and "sl_success" in d.columns:

            sl_val = d["sl_success"].iloc[i]

            if np.isfinite(sl_val):

                regime[i] = (
                    "ABS_CONT"
                    if sl_val == 1
                    else "ABS_REV"
                )

                continue


        # Fallback
        if dir_i == "BUY":

            regime[i] = (
                "ABS_REV"
                if fwd_price_chg[i] < 0
                else "ABS_CONT"
            )

        else:

            regime[i] = (
                "ABS_CONT"
                if fwd_price_chg[i] > 0
                else "ABS_REV"
            )


    d["regime"] = regime

    return d


# ═══════════════════════════════════════════════════════════════
# APPLY TO ALL DATA
# ═══════════════════════════════════════════════════════════════

all_enriched = {}

for instr, hd in all_labeled.items():

    all_enriched[instr] = {}

    for horizon, df in hd.items():

        if len(df) > max(10, FW + 5):

            all_enriched[instr][horizon] = enrich(
                df,
                fw=FW,
                vol_win=VOL_WIN
            )


print("Enrichment done.")


# ═══════════════════════════════════════════════════════════════
# MERGE ALL INSTRUMENTS
# ═══════════════════════════════════════════════════════════════

big = pd.concat(
    [
        df
        for hd in all_enriched.values()
        for df in hd.values()
    ],
    ignore_index=True
)


print(f"Combined DataFrame: {big.shape}")


# Show new features
print(
    big[
        [
            "instrument",
            "horizon",
            "regime",

            "delta_persist_past_3",
            "delta_persist_past_5",
            "delta_persist_past_10",

            "cumulative_delta_pressure_3",
            "cumulative_delta_pressure_5",
            "cumulative_delta_pressure_10",

            "delta_persist",       # FUTURE — analysis only
            "lambda_shift",        # FUTURE — analysis only
            "delta_accel"          # FUTURE — analysis only
        ]
    ].head(20)
)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde


def plot_feature_distribution(
    df,
    feature,
    regime_a="PURE_SELLING",
    regime_b="ABS_REV",
    horizon="1d",
    bins=20,
    kde=True,
    show_mean=True,
    show_median=False,
    figsize=(9, 5),
    title=None,
):
    """
    Compare the distribution of ANY feature between two regimes.

    Example:
        plot_feature_distribution(
            big,
            feature="delta_persist_past_5"
        )

    Works with:
        delta_persist_past_3
        delta_persist_past_5
        delta_persist_past_10
        cumulative_delta_pressure_3
        cumulative_delta_pressure_5
        cumulative_delta_pressure_10
        absorption_strength
        absorption_efficiency
        kyle_lambda
        rvol
        delta_accel
        etc.
    """

    # ---------------------------------------------------------
    # Filter horizon
    # ---------------------------------------------------------

    d = df[df["horizon"] == horizon].copy()

    # ---------------------------------------------------------
    # Select the two regimes
    # ---------------------------------------------------------

    a = (
        d.loc[d["regime"] == regime_a, feature]
        .replace([np.inf, -np.inf], np.nan)
        .dropna()
        .to_numpy()
    )

    b = (
        d.loc[d["regime"] == regime_b, feature]
        .replace([np.inf, -np.inf], np.nan)
        .dropna()
        .to_numpy()
    )

    if len(a) < 2 or len(b) < 2:
        print("Not enough observations.")
        print(f"{regime_a}: {len(a)}")
        print(f"{regime_b}: {len(b)}")
        return

    # ---------------------------------------------------------
    # Plot
    # ---------------------------------------------------------

    fig, ax = plt.subplots(figsize=figsize)

    # Common range
    all_values = np.concatenate([a, b])

    xmin = np.nanmin(all_values)
    xmax = np.nanmax(all_values)

    if xmin == xmax:
        xmin -= 1
        xmax += 1

    xs = np.linspace(xmin, xmax, 300)

    # ---------------------------------------------------------
    # Histogram
    # ---------------------------------------------------------

    ax.hist(
        a,
        bins=bins,
        density=True,
        alpha=0.20,
        label=f"{regime_a} histogram"
    )

    ax.hist(
        b,
        bins=bins,
        density=True,
        alpha=0.20,
        label=f"{regime_b} histogram"
    )

    # ---------------------------------------------------------
    # KDE
    # ---------------------------------------------------------

    if kde:

        for arr, label in [
            (a, f"{regime_a} (n={len(a)})"),
            (b, f"{regime_b} (n={len(b)})")
        ]:

            # KDE needs some variation
            if np.std(arr) > 1e-12:

                density = gaussian_kde(
                    arr,
                    bw_method="scott"
                )

                ys = density(xs)

                ax.plot(
                    xs,
                    ys,
                    linewidth=2,
                    label=label
                )

    # ---------------------------------------------------------
    # Means
    # ---------------------------------------------------------

    if show_mean:

        ax.axvline(
            np.mean(a),
            linestyle="--",
            linewidth=1.5,
            label=f"{regime_a} mean = {np.mean(a):.3f}"
        )

        ax.axvline(
            np.mean(b),
            linestyle="--",
            linewidth=1.5,
            label=f"{regime_b} mean = {np.mean(b):.3f}"
        )

    # ---------------------------------------------------------
    # Medians
    # ---------------------------------------------------------

    if show_median:

        ax.axvline(
            np.median(a),
            linestyle=":",
            linewidth=1.5
        )

        ax.axvline(
            np.median(b),
            linestyle=":",
            linewidth=1.5
        )

    # ---------------------------------------------------------
    # Labels
    # ---------------------------------------------------------

    ax.set_xlabel(feature)
    ax.set_ylabel("Density")

    if title is None:

        title = (
            f"{feature}\n"
            f"{regime_a} vs {regime_b} | horizon={horizon}"
        )

    ax.set_title(
        title,
        fontsize=12,
        fontweight="bold"
    )

    ax.legend(fontsize=8)

    ax.grid(
        alpha=0.15,
        linestyle="--"
    )

    plt.tight_layout()
    plt.show()

    # ---------------------------------------------------------
    # Print statistics
    # ---------------------------------------------------------

    print("\n" + "=" * 60)
    print(f"Feature: {feature}")
    print(f"Horizon: {horizon}")
    print("=" * 60)

    print(
        f"\n{regime_a}"
        f"\n  n      = {len(a)}"
        f"\n  mean   = {np.mean(a):.5f}"
        f"\n  median = {np.median(a):.5f}"
        f"\n  std    = {np.std(a, ddof=1):.5f}"
    )

    print(
        f"\n{regime_b}"
        f"\n  n      = {len(b)}"
        f"\n  mean   = {np.mean(b):.5f}"
        f"\n  median = {np.median(b):.5f}"
        f"\n  std    = {np.std(b, ddof=1):.5f}"
    )
```
